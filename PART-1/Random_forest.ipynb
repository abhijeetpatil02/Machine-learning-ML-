```python
import pandas as pd
from sklearn.ensemble import RandomForestRegressor
```

```python

# 1. Create the historical training dataset
data = {
    'Age': [25, 65, 45, 30, 70, 50, 35, 60],
    'Days_Stayed': [2, 7, 4, 1, 10, 5, 3, 8],
    'ICU_Room': [0, 1, 0, 0, 1, 1, 0, 1],
    'Total_Cost': [15000, 95000, 35000, 10000, 140000, 85000, 22000, 110000]
}

df = pd.DataFrame(data)
X = df[['Age', 'Days_Stayed', 'ICU_Room']]
y = df['Total_Cost']

# 2. Train the Random Forest Regressor
model = RandomForestRegressor(n_estimators=100, random_state=42)
model.fit(X, y)
print("Model trained successfully and ready for your inputs!")
print("=" * 45 + "\n")

# 3. Interactive Input Section
try:
    # Ask the user for inputs via the terminal
    user_age = int(input("Enter Patient's Age (e.g., 45): "))
    user_days = int(input("Enter Number of Days Stayed (e.g., 5): "))
    user_icu = int(input("Is it an ICU Room? (Type 1 for Yes, 0 for No): "))
    
    # Validation check for ICU input
    if user_icu not in [0, 1]:
        print("\n Error: For ICU Room, you must enter either 1 (Yes) or 0 (No).")
    else:
        # Create a new DataFrame using the user's specific inputs
        custom_patient_data = pd.DataFrame([{
            'Age': user_age, 
            'Days_Stayed': user_days, 
            'ICU_Room': user_icu
        }])

        # Predict using our trained model
        predicted_cost = model.predict(custom_patient_data)

        # Display the result nicely
        room_type = "ICU Room" if user_icu == 1 else "General Ward"
        print("\n" + "-" * 35)
        print("📋 PREDICTION RESULTS")
        print("-" * 35)
        print(f"Patient Profile : Age {user_age}, Staying {user_days} days in {room_type}")
        print(f"Estimated Cost  : ₹{predicted_cost[0]:,.2f}")
        print("-" * 35)

except ValueError:
    print("\n Error: Please enter valid numbers only.")
```
