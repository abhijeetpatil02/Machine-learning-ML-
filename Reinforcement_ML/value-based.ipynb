```python
class ValueBasedAgent:
    def __init__(self):
        # The AI's memory (Q-Table) starting with 0 predicted points
        self.action_scores = {"Left": 0.0, "Right": 0.0}
        self.learning_rate = 0.5  # How fast it updates its memory

    def choose_action(self):
        # Look at the scores and pick the highest one
        chosen = max(self.action_scores, key=self.action_scores.get)
        return chosen

    def learn(self, action, reward):
        # Update the score formula: Old Score + (Learning Rate * Reward)
        self.action_scores[action] += self.learning_rate * reward


# --- RUNNING THE SIMULATION ---
agent = ValueBasedAgent()

print("--- BEFORE TRAINING ---")
print(f"AI Memory Scores: {agent.action_scores}")
print(f"AI chooses to go: {agent.choose_action()}\n")

# Scenario: The AI tries going "Right" and hits the jackpot!
print("-> AI goes Right and gets a +100 reward!")
agent.learn(action="Right", reward=100)

# Scenario: The AI tries going "Left" and bumps a wall
print("-> AI goes Left and gets a -10 reward!\n")
agent.learn(action="Left", reward=-10)

print("--- AFTER TRAINING ---")
print(f"AI Memory Scores: {agent.action_scores}")
print(f"AI will now ALWAYS choose to go: {agent.choose_action()}")
```
